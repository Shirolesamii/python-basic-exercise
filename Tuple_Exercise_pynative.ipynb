Exercise 11: Function Returning Tuple

```python
def get_min_max(numbers):
    if not numbers:
        return(None,None)
    min_val=min(numbers)
    max_val=max(numbers)
    return(min_val,max_val)

my_numbers=[10,5,20,2,15]
min_max_values=get_min_max(my_numbers)
print("original number:",my_numbers)
print("minimum and maximum values:",min_max_values)
```

Exercise 13: Removing Duplicates from Tuple

```python
my_tuple = (1, 2, 2, 3, 4, 4, 5)
set=set(my_tuple)
new=tuple(set)
print("unique tuple:",new)
```

```python
from collections import OrderedDict
unique_ordered_tuple = tuple(OrderedDict.fromkeys(my_tuple))
print(f"Tuple with unique elements (order preserved): {unique_ordered_tuple}")
```

Exercise 14: Filter Tuples

```python
students = [('Alice', 85), ('Bob', 92), ('Charlie', 78), ('David', 95)]
print("original list:",students)

Loop=[]
for student in students:
    if student[1]>=90:
        Loop.append(student)
print("students with scores above 90:",Loop)
```

```python
loop=[student for student in students if student[1] >= 90]
print("students with scores above 90:",loop)
```

Exercise 15: Map Tuples

```python
#using list comprehension
t = (1, 2, 3, 4)
print("original tuple:",t)
squar=[i*i for i in t]
print("squred tuple:",squar)
```

```python
# with map()
t = (1, 2, 3, 4)
squared_tuple_map=tuple(map(lambda x:x**2,t))
print("squared tuple:",squared_tuple_map)
```
