Exercise 5: Count all letters, digits, and special symbols from a given string

```python
str1 = "P@#yn26at^&i5ve"
letters=0
digits=0
symbols=0

for char in str1:
    if char.isalpha():
        letters+=1

    elif char.isdigit():
        digits+=1

    else:
        symbols+=1
print("letters=", letters , "Digits=",digits, "symbol=",symbols )
```

Exercise 9: Calculate the sum and average of the digits present in a string

```python
str1 = "PYnative29@#8496"
digits=0
total=0
cnt=0

for i in str1:
    if i.isdigit():
        print(i)
        total+=int(i)
        cnt += 1
avg=total/cnt
print('sum:',total, "average:",avg)
```

```python
import re
str = "PYnative29@#8496"
digits_list=[int(num) for num in re.findall(r'\d',str)]
print('Digits:',digits_list)

total = sum(digits_list)
avg=total/len(digits_list)
print('sum:',total, "average:",avg)
```

Exercise 15: Remove special symbols / punctuation from a string

```python
import string
str1 = "/*Jon is @developer & musician"

new_str=str1.translate(str.maketrans('','',string.punctuation))

print("new string is",new_str)
```

```python
import re

str1 = "/*Jon is @developer & musician"
print("Original string is ", str1)

res=re.sub(r'[^\w\s]','',str1)
print('New string is',res)
```

Exercise 17: Find words with both alphabets and numbers

```python
str1 = "Emma25 is Data scientist50 and AI Expert"
res=[]
temp=str1.split()

for item in temp:
    if any(char.isalpha() for char in item) and any(char.isdigit() for char in item):
        res.append(item)

print("displaying words with alphabet and numbers")
for i in res:
    print(i)
```
