```python
#13: Write a program to find the number of vowels, consonants, digits, and white space characters in a string.
str=input('enter your string:')
print("Enter your string:",str)
vowels_char="aeiouAEIOU"
vowels=0
consonant=0
digit=0
whitespace=0
for x in str:
    if x.isdigit():
        digit+=1
    
    elif x.isspace():
        whitespace+=1

    elif x.isalpha():
        if x in vowels_char:
            vowels+=1
        else:
            consonant+=1

print('vowels',vowels)
print('consonant:',consonant) 
print('digit:',digit)
print('whitespace',whitespace)
```

```python
#14. Write a Python program to count Uppercase, Lowercase, special character, and numeric values in a given string.
str=input("Enter your string:")
print("string:",str)

uppercase=0
lowercase=0
special_chars="@#$%^&*()/\!"
special_char=0
number=0

for char in str:
    if char.islower():
        lowercase+=1
    elif char.isupper():
        uppercase+=1
    elif char.isnumeric():
        number+=1

    elif char in special_chars:
        special_char+=1
    

    
print('uppercase',uppercase)
print('lowercase',lowercase)
print('special characters',special_char)
print('number:',number)
```

```python
#15. Write a program to make a new string with all the consonants deleted from the string "Hello, have a good day".
str=input('enter your string:')
print("Enter your string:",str)
vowels_char="aeiouAEIOU"

res=''.join((i for i in str if not i.isalpha() or i in vowels_char))

print('new string:',res)
```

```python
#16. Write a Python program to remove the nth index character from a non-empty string.
str=input('enter your string:')
print('string:',str)
n = int(input("Enter the index to remove: "))

if 0 <= n < len(str):  
    result = str[:n] + str[n+1:]  
    print("String after removal:", result)
else:
    print("Invalid index!")
```

```python
#18. Write a Python program to count the occurrences of each word in a given sentence

sent=input("Enter your sentence:")
print("Sentence:",sent)

words=sent.split()
w_count={}

for word in words:
    if word in w_count:
        w_count[word]+=1

    else:
        w_count[word]=1
w_count

for word,count in w_count.items():
    print(f"{word} : {count}")
```

```python
#19:How do you count the occurrence of a given character in a string?
str=input("Enter your string:")
print("String:",str)
for char in set(str):
    count=str.count(char)
    print(f"{char}:{count}")
```
